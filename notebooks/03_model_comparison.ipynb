# 03. Model Comparison & Leaderboard

Compares statistical models and feature-based ML models across backtest metrics.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

leaderboard = pd.read_csv("../reports/metrics/model_comparison_leaderboard.csv")
display(leaderboard)

plt.figure(figsize=(10, 5))
sns.barplot(data=leaderboard, x="model_name", y="rmse_mean", hue="horizon_hours")
plt.title("Mean RMSE across Models and Forecast Horizons")
plt.ylabel("RMSE (kWh)")
plt.show()
```
