# 04. Walk-Forward Backtesting Deep Dive

Detailed evaluation of expanding-window folds and residual stability over time.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

backtests_dir = Path("../reports/backtests")
fold_files = list(backtests_dir.glob("*.parquet"))
print(f"Found {len(fold_files)} backtest prediction files.")

if fold_files:
    sample_df = pd.read_parquet(fold_files[0])
    sample_df.plot(x="timestamp", y=["energy_demand_kwh", "forecast_kwh"], figsize=(12, 5))
    plt.title(f"Backtest Predictions: {fold_files[0].stem}")
    plt.ylabel("Energy Demand (kWh)")
    plt.show()
```
